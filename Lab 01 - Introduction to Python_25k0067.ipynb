```python
#C5
numbers = [3, 2, 3, 4, 1, 2, 3]

while True:

    print(
        "1. Add a number\n"
        "2. Remove a number\n"
        "3. Show the list\n"
        "4. Show statistics\n"
        "5. Exit"
    )

    user = int(input("Enter your choice: "))

    if user == 5:
        break

    elif user == 1:
        num = int(input("Enter number: "))
        numbers.append(num)

    elif user == 2:
        num = int(input("Number to remove: "))

        if num in numbers:
            numbers.remove(num)
        else:
            print("Number not found")

    elif user == 3:
        print(numbers)

    elif user == 4:
        print("Max:", max(numbers))
        print("Min:", min(numbers))
        print("Average:", sum(numbers) / len(numbers))

        print("Even and odd numbers:")

        for num in numbers:
            if num % 2 == 0:
                print(num, "Even")
            else:
                print(num, "Odd")

    else:
        print("Invalid choice")
```

```python
#C4
record = {}

n = int(input("How many subjects? "))

# Input subjects and marks
for i in range(n):
    subject = input("Enter subject name: ")
    marks = int(input("Enter marks: "))

    record[subject] = marks


# Display marks and grade
for subject, marks in record.items():

    if marks >= 80:
        grade = "A"
    elif marks >= 70:
        grade = "B"
    elif marks >= 60:
        grade = "C"
    elif marks >= 50:
        grade = "D"
    else:
        grade = "F"

    print(subject, marks, grade)


# Average
average = sum(record.values()) / len(record)

print("Average:", average)


# Highest and lowest
highest = max(record, key=record.get)
lowest = min(record, key=record.get)

print("Highest:", highest, record[highest])
print("Lowest:", lowest, record[lowest])
```

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
op = input("Enter operation (+, -, *, /): ")

if op == "+":
    print("Result:", a - (-b))

elif op == "-":
    print("Result:", a - b)

elif op == "*":
    print("Result:", a * b)

elif op == "/":
    if b != 0:
        print("Result:", a / b)
    else:
        print("Cannot divide by zero")

else:
    print("Invalid operation")
```
