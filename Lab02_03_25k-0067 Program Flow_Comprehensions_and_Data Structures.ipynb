## 1. Conditional Statements in Depth

### 1.1 The `elif` chain stops at the first true condition

Everything below the first true branch is skipped — which is why the **order** of the branches
matters. Run the cell and explain to yourself why `85` is reported as a plain pass.

````python
marks = 85

if marks >= 50:
    grade = 'Pass'            # true first, so the chain stops here
elif marks >= 80:
    grade = 'Distinction'     # never reached, even though 85 >= 80

print(grade)
````

````python
# The fix: test the narrowest / highest condition first
marks = 85

if marks >= 80:
    grade = 'Distinction'
elif marks >= 50:
    grade = 'Pass'
else:
    grade = 'Fail'

print(grade)
````

### 1.2 Nesting versus a combined condition

````python
marks, attendance = 85, 60

# nested: the inner question only matters once the outer one is answered
if attendance >= 75:
    if marks >= 80:
        print("Distinction")
    else:
        print("Pass")
else:
    print("Not eligible - attendance short")

# the same two independent conditions, in one level
if attendance >= 75 and marks >= 80:
    print("Distinction")
else:
    print("Not a distinction")
````

### 1.3 The conditional expression

`A if condition else B` produces a **value**. Use it when both outcomes are short — you will
need it again inside comprehensions.

````python
n = 7

# long form
if n % 2 == 0:
    kind = "even"
else:
    kind = "odd"
print(n, "is", kind)

# one-line conditional expression - exactly the same result
kind = "even" if n % 2 == 0 else "odd"
print(n, "is", kind)
````

### 1.4 Truthiness

`0`, `0.0`, `""`, `[]`, `{}`, `set()` and `None` are false; everything else is true.

````python
items = []

if items:
    print("the list has", len(items), "items")
else:
    print("the list is empty")

print(bool(0), bool(""), bool([]), bool({}), bool(None))
print(bool(-1), bool("0"), bool([0]))     # not empty -> all true
````

## 2. Loops in Depth

### 2.1 `for` when the passes are known, `while` when they are not

````python
# for: the sequence fixes the number of passes
for subject in ['Physics', 'Chemistry', 'Maths']:
    print(subject)

# while: the data decides how many passes are needed
total, count = 0, 0
while total < 100:
    total += 15
    count += 1
print("took", count, "passes to pass 100")
````

### 2.2 `enumerate()` — the index and the item together

````python
subjects = ['Physics', 'Chemistry', 'Maths']

for i in range(len(subjects)):          # works, but clumsy
    print(i, subjects[i])

print("---")
for i, subject in enumerate(subjects):  # preferred
    print(i, subject)

print("---")
for rank, subject in enumerate(subjects, start=1):
    print(rank, subject)
````

### 2.3 `zip()` — walking two sequences side by side

````python
subjects = ['Physics', 'Chemistry', 'Maths']
marks    = [78, 45, 90]

for subject, mark in zip(subjects, marks):
    print(subject, "->", mark)

print(dict(zip(subjects, marks)))       # a quick way to build a dictionary

# zip stops at the shortest sequence
print(list(zip([1, 2, 3, 4], ['a', 'b'])))
````

### 2.4 The `else` clause of a loop

It runs when the loop finishes **normally**, and is skipped when the loop is left by `break`.
That makes it the tidy way to say "I searched everything and found nothing".

````python
numbers = [4, 9, 16, 25]

for target in (16, 7):
    for n in numbers:
        if n == target:
            print("found", target)
            break
    else:
        print(target, "is not in the list")
````

## 3. `break`, `continue` and `pass`

| Keyword | Effect |
|---|---|
| `break` | Leaves the innermost loop at once; the loop's `else` is skipped |
| `continue` | Abandons the rest of this pass and goes back to the loop header |
| `pass` | Does nothing — a placeholder that keeps an empty block valid |

````python
for n in range(1, 12):
    if n % 2 == 0:
        continue        # skip the even numbers
    if n > 7:
        break           # stop once past 7
    print(n, end=' ')
print("\ndone")
````

````python
# break leaves ONE level of loop only
for i in range(1, 4):
    for j in range(1, 4):
        if j == 2:
            break       # ends the inner loop, not the outer one
        print(i, j)
````

````python
# pass: Python refuses an empty block, so this keeps the code runnable
for n in range(3):
    pass

if False:
    pass

print("both blocks were valid Python")
````

## 4. The Five Standard Loop Patterns

Almost every loop in this course is one of these shapes.

````python
# 1. ACCUMULATING - start from a neutral value and update once per pass
marks = [78, 45, 90, 62]

total = 0
for m in marks:
    total += m
print("total:", total, "| average:", total / len(marks))
````

````python
# 2. COUNTING - advance the counter only when a condition holds
passed = 0
for m in marks:
    if m >= 50:
        passed += 1
print(passed, "of", len(marks), "passed")
````

````python
# 3. SEARCHING - stop at the first hit; the loop else reports failure
subjects = ['Physics', 'Chemistry', 'Maths']
wanted = 'Maths'

for i, subject in enumerate(subjects):
    if subject == wanted:
        print(wanted, "found at index", i)
        break
else:
    print(wanted, "not found")
````

````python
# 4. TRACKING A MAXIMUM - assume the first is best, then let the loop correct you
record = {'Physics': 78, 'Chemistry': 45, 'Maths': 90}

best_subject, best_mark = None, -1
for subject, mark in record.items():
    if mark > best_mark:
        best_subject, best_mark = subject, mark

print("highest:", best_subject, best_mark)
print("shortcut:", max(record, key=record.get))
````

````python
# 5. SENTINEL-CONTROLLED INPUT - keep reading until an agreed stopping value
# (fixed values here so the notebook runs; the real version is commented)

entries = ["12", "7", "0"]      # what the user would have typed
numbers = []

for entry in entries:           # while True: entry = input("Enter a number (0 to finish): ")
    value = int(entry)
    if value == 0:
        break
    numbers.append(value)

print("you entered", len(numbers), "numbers:", numbers)
````

````python
# Nested loops: the inner loop runs completely for every pass of the outer loop
for i in range(1, 6):
    for j in range(i):
        print("*", end="")
    print()

print()
for i in range(1, 6):
    print("*" * i)          # the same pattern, using string repetition
````

## 5. Comprehensions

A comprehension is a single expression that builds a whole collection — Python's idiomatic
replacement for "make an empty list, loop, append".

```
[ n ** 2      for n in numbers      if n % 2 == 0 ]
  ^^^^^^      ^^^^^^^^^^^^^^^^      ^^^^^^^^^^^^^
  output        iteration            optional filter
```

````python
numbers = [1, 2, 3, 4, 5, 6]

# the loop you already know
squares = []
for n in numbers:
    squares.append(n ** 2)
print(squares)

# exactly the same result as a comprehension
squares = [n ** 2 for n in numbers]
print(squares)
````

### 5.1 Filtering with `if`

````python
numbers = [1, 2, 3, 4, 5, 6]
words = ['python', 'ai', 'lab', 'comprehension', 'data']

print([n ** 2 for n in numbers if n % 2 == 0])
print([w for w in words if len(w) > 3])
print([w.upper() for w in words if len(w) > 3])
````

### 5.2 Choosing between two values

Note the position: `if ... else` **before** the `for` chooses a value for every item;
a bare `if` **after** the `for` filters items out.

````python
numbers = [1, 2, 3, 4, 5, 6]

labels = ["even" if n % 2 == 0 else "odd" for n in numbers]   # chooses
evens  = [n for n in numbers if n % 2 == 0]                   # filters

print(labels)
print(evens)
````

### 5.3 Nested comprehensions

````python
# a multiplication grid: three rows of four columns
grid = [[row * col for col in range(1, 5)] for row in range(1, 4)]
print(grid)

# flattening a list of lists - the two for clauses read in loop order
flat = [value for row in grid for value in row]
print(flat)
````

### 5.4 Dictionary and set comprehensions

````python
words = ['python', 'ai', 'lab']

print({w: len(w) for w in words})                 # dictionary comprehension
print({n: n ** 2 for n in range(1, 6)})

record = {'Physics': 78, 'Chemistry': 45}
print({mark: subject for subject, mark in record.items()})    # inverted
````

````python
marks = [78, 45, 90, 45, 78]

print({m for m in marks})                       # set comprehension -> unique
print({m // 10 * 10 for m in marks})            # unique tens bands
print({w[0] for w in ['ali', 'ahmed', 'bilal']})
# a set has no order: the printed order may differ from run to run
````

### 5.5 When *not* to use a comprehension

Keep the ordinary loop when the body needs several statements, or when you are printing rather
than collecting a value.

````python
# poor: a comprehension used only for its side effect - it builds a list of None
[print(n) for n in range(3)]
````

````python
# good: an ordinary loop
for n in range(3):
    print(n)
````

## 6. Lists

Ordered, changeable, duplicates allowed — the default choice for a sequence of values.

````python
L = [3, 1, 2]

L.append(4)          # one item at the end
L.insert(0, 9)       # at a position
L.extend([5, 6])     # every item of another sequence
print(L)

print("popped:", L.pop())    # removes AND returns the last item
L.remove(9)                  # removes the first occurrence of a VALUE
print(L, "| index of 2:", L.index(2), "| count of 3:", L.count(3))
````

````python
# append vs extend - the classic mix-up
a = [1, 2]
b = [1, 2]

a.append([3, 4])     # adds ONE item that happens to be a list
b.extend([3, 4])     # adds TWO items

print("append:", a, "-> length", len(a))
print("extend:", b, "-> length", len(b))
````

### 6.1 Sorting: `sort()` changes the list, `sorted()` returns a new one

````python
marks = [78, 45, 90, 62]

print(sorted(marks))                  # new list
print(sorted(marks, reverse=True))
print(marks)                          # original untouched

marks.sort()                          # in place; returns None
print(marks)

broken = [3, 1, 2].sort()             # the classic mistake
print("value of a sort() call:", broken)
````

### 6.2 Copying versus aliasing

Assigning a list to a second name does **not** copy it — both names point at the same object.

````python
original = [1, 2, 3]

alias = original          # same list, second name
copy  = original[:]       # independent copy (list(original) also works)

alias.append(99)

print("original:", original)
print("alias   :", alias)
print("copy    :", copy)
````

### 6.3 Slice assignment and two-dimensional lists

````python
items = [1, 2, 3, 4, 5]
items[1:3] = [20, 30]        # replace a whole slice
print(items)

matrix = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]

print(matrix[1])             # second row
print(matrix[1][2])          # row 1, column 2

for row in matrix:
    print(row, "-> sum", sum(row))
````

## 7. Tuples

Ordered like a list, but **immutable**. That restriction is the point: it protects fixed data
and makes a tuple usable where a list is not allowed.

````python
point   = (3, 4)
record  = ('22K-1234', 'Ali', 3.4)
single  = (5,)          # the comma makes it a tuple
notuple = (5)           # without the comma this is just an int

print(type(single), type(notuple))
print(len(record), record[1], record[-1])
````

````python
# packing and unpacking
record = ('22K-1234', 'Ali', 3.4)
roll, name, cgpa = record
print(roll, name, cgpa)

a, b = 10, 20
a, b = b, a             # pack on the right, unpack on the left
print(a, b)
````

````python
# a tuple can be a dictionary key; a list cannot
distances = {('Karachi', 'Lahore'): 1210,
             ('Karachi', 'Quetta'): 690}
print(distances[('Karachi', 'Lahore')])

# a list of tuples is the standard shape for tabular data
students = [('Ali', 78), ('Sara', 92), ('Bilal', 65)]
for name, mark in students:
    print(name, mark)
````

## 8. Dictionaries

Values stored against **keys** rather than positions. Keys are unique and immutable; insertion
order is preserved from Python 3.7 onwards.

````python
marks = {'Physics': 78, 'Maths': 90}

print(marks.get('Chemistry'))         # None instead of an error
print(marks.get('Chemistry', 0))      # a sensible default
print('Chemistry' in marks)           # tests KEYS, never values

marks['Chemistry'] = 45               # add
marks['Physics'] = 80                 # overwrite
marks.setdefault('English', 65)       # only if absent
marks.setdefault('Physics', 0)        # ignored - Physics already exists
print(marks)

print("removed:", marks.pop('English'))
print(len(marks), "subjects left")
````

````python
marks = {'Physics': 78, 'Chemistry': 45, 'Maths': 90}

for subject in marks:                 # keys by default
    print(subject, end=' ')
print()

for mark in marks.values():
    print(mark, end=' ')
print()

for subject, mark in marks.items():   # both, unpacked
    print(subject, "->", mark)

print("average:", sum(marks.values()) / len(marks))
print("best   :", max(marks, key=marks.get))
````

### 8.1 The frequency-counting pattern

The single most useful thing a dictionary does: take the count so far (defaulting to zero) and
store it back increased by one.

````python
sentence = "the lab is the first lab of the course"
counts = {}

for word in sentence.split():
    counts[word] = counts.get(word, 0) + 1

print(counts)
print("most frequent:", max(counts, key=counts.get))
print("seen more than once:", [w for w, c in counts.items() if c > 1])
````

### 8.2 Nested dictionaries and sorting by value

````python
students = {
    '22K-1234': {'name': 'Ali',  'marks': [78, 45, 90]},
    '22K-5678': {'name': 'Sara', 'marks': [88, 92, 95]},
}

for roll, info in students.items():
    average = sum(info['marks']) / len(info['marks'])
    print(roll, info['name'], round(average, 2))
````

````python
marks = {'Physics': 78, 'Chemistry': 45, 'Maths': 90}

for subject in sorted(marks, key=marks.get, reverse=True):
    print(subject, marks[subject])
````

## 9. Sets

Unordered, unique, not indexable. Two jobs: removing duplicates, and comparing collections.

````python
marks = [78, 45, 90, 45, 78]

unique = set(marks)
print(unique, "->", len(unique), "distinct values")

empty = set()                 # NOT {} - that is an empty dictionary
print(type(empty), type({}))

unique.add(60)
unique.discard(45)            # no complaint if the item is absent
print(unique, "| 78 present?", 78 in unique)
````

````python
section_a = {'Ali', 'Sara', 'Bilal', 'Hina'}
section_b = {'Sara', 'Hina', 'Usman'}

print("all students     :", section_a | section_b)   # union
print("in both sections :", section_a & section_b)   # intersection
print("only in A        :", section_a - section_b)   # difference
print("in exactly one   :", section_a ^ section_b)   # symmetric difference
print("B inside A?      :", section_b <= section_a)  # subset
````

````python
# What a set cannot do
S = {3, 1, 2}
print(sorted(S))        # sort it into a list when order matters

try:
    print(S[0])         # sets are not indexable
except TypeError as err:
    print("TypeError:", err)
````

## 10. Choosing the Right Structure

| | List | Tuple | Dictionary | Set |
|---|---|---|---|---|
| Written with | `[ ]` | `( )` | `{k: v}` | `{ }` / `set()` |
| Ordered | Yes | Yes | By insertion | No |
| Changeable | Yes | No | Yes | Yes |
| Duplicates | Allowed | Allowed | Keys unique | Removed |
| Accessed by | Index | Index | Key | Membership only |

Membership is the difference that matters most: `x in list` compares item after item, while
`x in set` and `key in dict` jump straight to the answer. On a few items you cannot tell; on
tens of thousands it is the difference between instant and slow.

````python
names = ['Ali', 'Sara', 'Bilal'] * 1000     # a long list with repeats
lookup = set(names)                          # build the set once

print(len(names), "entries ->", len(lookup), "distinct")
print('Sara' in names)      # scans the list
print('Sara' in lookup)     # one hash lookup - the shape to prefer
````

````python
#2
numbers = [12, 7, 40, 3, 25, 18, 7, 40]

words = ['python', 'ai', 'lab', 'data', 'comprehension']


 
odd_squares = []

for num in numbers:
    if num % 2 != 0:
        odd_squares.append(num ** 2)

print("1. Loop:", odd_squares)

odd_squares_comp = [num ** 2 for num in numbers if num % 2 != 0]

print("1. Comprehension:", odd_squares_comp)


 
long_words = []

for word in words:
    if len(word) > 3:
        long_words.append(word.upper())

print("2. Loop:", long_words)

long_words_comp = [word.upper() for word in words if len(word) > 3]

print("2. Comprehension:", long_words_comp)


 
word_lengths = {}

for word in words:
    word_lengths[word] = len(word)

print("3. Loop:", word_lengths)

word_lengths_comp = {word: len(word) for word in words}

print("3. Comprehension:", word_lengths_comp)
 

first_letters = set()

for word in words:
    first_letters.add(word[0])

print("4. Loop:", first_letters)

first_letters_comp = {word[0] for word in words}

print("4. Comprehension:", first_letters_comp)



labels = []

for num in numbers:
    if num % 2 == 0:
        labels.append("even")
    else:
        labels.append("odd")

print("5. Loop:", labels)

labels_comp = ["even" if num % 2 == 0 else "odd" for num in numbers]

print("5. Comprehension:", labels_comp)


````

````python
#3
d1 = {"Marks": [12, 32, 43, 54, 65], "name": "hussain", "age": 23}
d2 = {"Marks": [12, 32, 43, 54, 65], "name": "MR hussain", "age": 26}
d3 = {"Marks": [12, 32, 43, 64, 95], "name": "hussain Al", "age": 27}
d4 = {"Marks": [12, 32, 43, 54, 75], "name": "hussain Qamari", "age": 29}
d5 = {"Marks": [12, 42, 43, 54, 65], "name": "hussain al qamri", "age": 23}

students = [d1, d2, d3, d4, d5]



for s in students:
    total_marks = sum(s["Marks"])
    avg_marks = total_marks / len(s["Marks"])
    print(s["name"], "-> Total:", total_marks, "| Average:", avg_marks)

top_score = -1
topper_name = ""

for s in students:
    total_marks = sum(s["Marks"])
    if total_marks > top_score:
        top_score = total_marks
        topper_name = s["name"]

print("Topper is:", topper_name, "with total marks:", top_score)


below_50_students = []

for s in students:
    avg_marks = sum(s["Marks"]) / len(s["Marks"])
    if avg_marks < 50:
        below_50_students.append(s["name"])

print("\n--- 3. Students with Average Below 50 ---")
print(below_50_students)


avg_name_list = []

for s in students:
    avg_marks = sum(s["Marks"]) / len(s["Marks"])
    student_tuple = (avg_marks, s["name"])
    avg_name_list.append(student_tuple)


sorted_list = sorted(avg_name_list, reverse=True)

for item in sorted_list:
    print(item)
````

````python
sentence = input("Enter the sentence: ")

words = sentence.lower().split()


print("Total words:", len(words))

distinct = set(words)
print("Distinct words:", len(distinct))

 
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print("Frequency:", frequency)

# Repeated words
repeated = []

for word in frequency:
    if frequency[word] > 1:
        repeated.append(word)

print("Repeated words:", repeated)

# Longest word
longest = max(words, key=len)
print("Longest word:", longest)

# Word lengths
word_lengths = {}

for word in frequency:
    word_lengths[word] = len(word)

print("Word lengths:", word_lengths)

# Function to get frequency
def get_count(item):
    return item[1]

# Sort frequency from highest to lowest
sorted_frequency = sorted(
    frequency.items(),
    key=get_count,
    reverse=True
)

print("Frequency table:")

for word, count in sorted_frequency:
    print(word, ":", count)

    
````
